### **Data Preparation**

I have decided not to replace ** with a zero value. The BLS dataset's documentation specifically states that "**" this means that the employment estimate is not available. It does not mean no employment exists. If treated with zero, columns like "TOT_EMP" will take it as no employment exists. These symbols cannot be treated as numerical values as well. So, they have to be converted to numerical values.  

When visualisation requires the columns and rows with these symbols they should be coverted, if not they should be replaced, dropped or dealt with. Except that the dataset will remain unchanged.

## **2.National Ranking Chart**  

### 2.1 Prepare data for National Level

```python
# Check data types to filter the National Values
print(df[["AREA_TYPE", "I_GROUP", "OWN_CODE", "O_GROUP"]].dtypes)  

print("Example category values:")
display(df[["AREA_TYPE", "I_GROUP", "OWN_CODE", "O_GROUP"]].head())
```

```python
# Create nat_detailed as a copy 
nat_detailed = df[
    (df["AREA_TYPE"] == 1) &
    (df["I_GROUP"] == "cross-industry") &
    (df["OWN_CODE"] == 1235) &
    (df["O_GROUP"] == "detailed") &
    (df["OCC_CODE"].str.startswith("15-", na=False))
    ].copy()
```

```python
# Select the required columns
nat_detailed = nat_detailed[
    ["OCC_CODE", "OCC_TITLE", "TOT_EMP"]
    ].copy()
```

```python
# Convert employment to numeric
nat_detailed["TOT_EMP"] = pd.to_numeric(
    nat_detailed["TOT_EMP"],
    errors="coerce"
)
```

```python
# Sort by employment
nat_detailed = nat_detailed.sort_values(
    "TOT_EMP",
    ascending=False
).reset_index(drop=True)
```

```python
# Verify the dataset
print("Dataset shape:", nat_detailed.shape)

print("Data Types:")
print(nat_detailed.dtypes)

print("Missing Values:")
print(nat_detailed.isna().sum())

print("Top 10 Occupations:")
print(nat_detailed.head(10))
```

## 2.2 National Employment Ranking Chart Across All Occupation Related To Computer And Maths

```python
# Highlighing the selected three occupation

selected_occupations = [
    "Computer and Information Research Scientists",
    "Data Scientists",
    "Software Developers"
]

bar_colors = [
    "#006684" if occupation in selected_occupations
        else "#B8C9CE"
            for occupation in nat_detailed["OCC_TITLE"]
]

# Create the chart
plt.figure(figsize=(13,10))

# Create the horizontal bars
plt.barh(nat_detailed["OCC_TITLE"],
         nat_detailed["TOT_EMP"],
        color = bar_colors
        )

# Display the highest employments at the top
plt.gca().invert_yaxis()

plt.title(
    "National Employment Rankings Of Computer And Mathematical Occupations",
    fontsize = 15,
    fontweight = "bold",
    pad= 18
)
plt.xlabel("Total Employment")
plt.ylabel("Occupation")

# Makes employment easy to compare
plt.grid(axis = "x", alpha = 0.25)

# Adding values to the bars
plt.bar_label(
    bars,
    labels=[
        f"{value:,.0f}"
        for value in nat_detailed["TOT_EMP"]
    ],
    padding = 4,
    fontsize = 8
)

# Prevents any labels from being cut off
plt.tight_layout()

plt.show()
```

### Objective  
The objective of this horizontal bar chart is to compare the National Employment Levels of 21 detailed Computer and Maths Occupations in the US. It also specifically highlights the ranking of the three selected occupations such as Data Scientists, Software Developers and Computer and Research Scientists. They are highlighted as they will be used for further analysis and visualisations.

### Why choose a Horizontal Bar  
It was selected as it allows easy comparison of employment figures across multiple occupations. The horizontal layout makes oit easier to accomodate longer names and also understanding the long list of occupational information easier. It makes volume differences for such long list of occupations easier and also does a better job at displaying them in descending order.  

### What We Learned  
The visualised data shows us that Software Developers have the highest employment among the 21 occupations with 1,656,880 making them rank first. Data Scientist holds the rank 7 with 192,710 workers. Computer and Research Scientist ranks 17th with 35,210 workers making them the one with least jobs among the selected occupations.  

The results suggests that there is a substancial difference in the employment level in these three areas.
