## **2.National Ranking Chart**  

### 2.1 Prepare data for National Level

```python
# Check data types to filter the National Values
print(df[["AREA_TYPE", "I_GROUP", "OWN_CODE", "O_GROUP"]].dtypes)  

print("Example category values:")
display(df[["AREA_TYPE", "I_GROUP", "OWN_CODE", "O_GROUP"]].head())
```

```python
# Create nat_detailed as a copy 
nat_detailed = df[
    (df["AREA_TYPE"] == 1) &
    (df["I_GROUP"] == "cross-industry") &
    (df["OWN_CODE"] == 1235) &
    (df["O_GROUP"] == "detailed") &
    (df["OCC_CODE"].str.startswith("15-", na=False))
    ].copy()
```

```python
# Select the required columns
nat_detailed = nat_detailed[
    ["OCC_CODE", "OCC_TITLE", "TOT_EMP"]
    ].copy()
```

```python
# Convert employment to numeric
nat_detailed["TOT_EMP"] = pd.to_numeric(
    nat_detailed["TOT_EMP"],
    errors="coerce"
)
```

```python
# Sort by employment
nat_detailed = nat_detailed.sort_values(
    "TOT_EMP",
    ascending=False
).reset_index(drop=True)
```

```python
# Verify the dataset
print("Dataset shape:", nat_detailed.shape)

print("Data Types:")
print(nat_detailed.dtypes)

print("Missing Values:")
print(nat_detailed.isna().sum())

print("Top 10 Occupations:")
print(nat_detailed.head(10))
```

### 2.2 National Employment Ranking Chart Across All Occupation Related To Computer And Maths

```python
# Highlighing the selected three occupation

selected_occupations = [
    "Computer and Information Research Scientists",
    "Data Scientists",
    "Software Developers"
]

bar_colors = [
    "#006684" if occupation in selected_occupations
        else "#B8C9CE"
            for occupation in nat_detailed["OCC_TITLE"]
]

# Create the chart
plt.figure(figsize=(13,10))

# Create the horizontal bars
bars = plt.barh(
    nat_detailed["OCC_TITLE"],
    nat_detailed["TOT_EMP"],
    color = bar_colors
)

# Display the highest employments at the top
plt.gca().invert_yaxis()

plt.title(
    "National Employment Rankings Of Computer And Mathematical Occupations",
    fontsize = 15,
    fontweight = "bold",
    pad= 18
)
plt.xlabel("Total Employment")
plt.ylabel("Occupation")

# Makes employment easy to compare
plt.grid(axis = "x", alpha = 0.25)

# Adding values to the bars
plt.bar_label(
    bars,
    labels=[
        f"{value:,.0f}"
        for value in nat_detailed["TOT_EMP"]
    ],
    padding = 4,
    fontsize = 8
)

# Prevents any labels from being cut off
plt.tight_layout()

plt.show()
```

### Objective  
The objective of this horizontal bar chart is to compare the National Employment Levels of 21 detailed Computer and Maths Occupations in the US. It also specifically highlights the ranking of the three selected occupations such as Data Scientists, Software Developers and Computer and Research Scientists. They are highlighted as they will be used for further analysis and visualisations.

### Why choose a Horizontal Bar  
It was selected as it allows easy comparison of employment figures across multiple occupations. The horizontal layout makes oit easier to accomodate longer names and also understanding the long list of occupational information easier. It makes volume differences for such long list of occupations easier and also does a better job at displaying them in descending order.  

### What We Learned  
The visualised data shows us that Software Developers have the highest employment among the 21 occupations with 1,656,880 making them rank first. Data Scientist holds the rank 7 with 192,710 workers. Computer and Research Scientist ranks 17th with 35,210 workers making them the one with least jobs among the selected occupations.  

The results suggests that there is a substancial difference in the employment level in these three areas. Software Developers have nearly 8.6 times jobs than Data Scientists and around 47 times that of Computer and Research Scientists. The US Employment Data of May 2023 shows us that the employment scale of Software Developers is much more than that of the other two selected occupations.

### 3.2 Median Annual Wage Variation Across States

```python
# Prepare the chart data

# Remove records where the wage is missing
box_data = state_data.dropna(
    subset = ["A_MEDIAN"]
).copy()

print("Box plot dataset shape:", box_data.shape)

print("\n\nMissing median wages:")
print(box_data["A_MEDIAN"].isna().sum())
```

```python
# Create the chart
plt.figure(figsize = (11, 7))

sns.boxplot(
    data = box_data,
    x = "OCC_TITLE",
    y = "A_MEDIAN",
    hue = "OCC_TITLE",
    palette = {
        "Computer and Information Research Scientists": "blue",
        "Data Scientists": "green",
        "Software Developers": "orange"
    },
    legend = False
)

# Add the title
plt.title(
    "Variation in Median Annual Wages Across 50 States And Washington DC By Occupation",
    fontsize = 15,
    fontweight = "bold",
    pad = 15
)

# Label the axis
plt.xlabel("Occupation")
plt.ylabel("Median Annual Wage ($)")

# Rotate the labels to avoid overlap
plt.xticks(rotation = 15)
plt.grid(axis = "y", alpha = 0.25)

# Prevents any labels from being cut off
plt.tight_layout()

plt.show()
```

### 3.3 State Level Employment versus Median Annual Wages

```python
# Prepare the data for the chart

# Remove the missing median values
scatter_data = state_data.dropna(
    subset = ["TOT_EMP"]
).copy()

# Remove invalid values like zero or negative values
scatter_data = scatter_data[
    (scatter_data["TOT_EMP"] > 0) &
    (scatter_data["A_MEDIAN"] > 0)
    ].copy()

print("Scatter plot dataset shape:", scatter_data.shape)

print("\n\nMissing Values:")
print(scatter_data[["TOT_EMP", "A_MEDIAN"]].isna().sum())
```

```python
# Create the chart
plt.figure(figsize =(11, 7))

sns.scatterplot(
    data = scatter_data,
    x = "TOT_EMP",
    y = "A_MEDIAN",
    hue = "OCC_TITLE",
    style = "OCC_TITLE",
    s = 100,
    palette = {
        "Computer and Information Research Scientists": "blue",
        "Data Scientists": "green",
        "Software Developers": "red"
    }
)

# Stopping small values getting squeezed to the left
plt.xscale("log")

# Chart title
plt.title(
    "State Employment vs Median Annual Wages",
    fontsize = 15,
    fontweight = "bold",
    pad = 15
)

# Labelling the axises
plt.xlabel("Total Employment (Logarithmic Scale)")
plt.ylabel("Median Annual Wage ($)")

# Grid lines
plt.grid(alpha = 0.25)

# Shows which colour and marker corresponds to each occupation
plt.legend(title = "Occupation")

plt.tight_layout()
plt.show()
```

### Objective  
The objective of this visualisation ia to explore the relationship between the employment levels and median annual wages for the selected three occupations across the US geographical areas.  

### Why Use Scatter Plot  
A Scatter Plot was used because it allows two distinct numerical variables to be compared at the same time. Each point will represent a geographical area, while different colours and markers represent the three selected occupations. A logarithmic scale was applied to the employment axis of this scatter plot because employment scales varies a lot among the occupations.

### 3.5 Prepare The Data For Heatmap

```python
# Create the dataframe for the heatmap
heatmap_data = state_data[
    [
        "AREA_TITLE",
        "PRIM_STATE",
        "OCC_CODE",
        "OCC_TITLE",
        "LOC_QUOTIENT"
    ]
    ].copy()
```

```python
# Check the dataset
print("\n\nMissing values:")
print(heatmap_data.isna().sum())

print("\n\nNumber of geographical areas:")
print(heatmap_data["AREA_TITLE"].nunique())

print("\n\nDuplicate occupation-area combinations:")
print(
    heatmap_data.duplicated(
        subset = ["AREA_TITLE", "OCC_TITLE"]
    ).sum()
)

display(heatmap_data.head(10))
```

```python
# Create the heatmap matrix
heatmap_matrix = heatmap_data.pivot(
    index = "AREA_TITLE",
    columns = "OCC_TITLE",
    values = "LOC_QUOTIENT"
)

print("Heatmap matrix shape:")
print(heatmap_matrix.shape)

print("\n\nMissing values in heatmap:")
print(heatmap_matrix.isna().sum())

display(heatmap_matrix.head(10))
```

```python
# Create the heatmap
plt.figure(figsize = (12,18))

sns.heatmap(
    heatmap_matrix,
# Show numbers inside the cell
    annot = True,
# Format numbers to show decimals to two places
    fmt = ".2f",
# Colour Scheme
    cmap = "RdYlGn",
# Centre at 1
    center = 1,
# Add borders between cells to make them easier to read
    linewidth = 0.5,
# Colour bar label
    cbar_kws = {
        "label": "Location Quotient"
    }
)

plt.title(
    "Geographical Concentration Of Selected Occupation Across the USA",
    fontsize = 15,
    fontweight = "bold",
    pad = 15
)

plt.xlabel("Occupation")
plt.ylabel("Geographical Areas")

# Tilt the lables
plt.xticks(rotation = 15, ha = "right")
plt.yticks(rotation = 0)

plt.tight_layout()
plt.show()
```

### Objective  

This visualisation has been done to compare the geographical concentration of the three selected occupations across the 50 US States and Washington, D.C.  

### Why Choose Heatmap  

A heatmap was selected because it allows the comparison of the multiple occupations, in this case the selected occupations across numerous gerographical location in a single chart. The different colours here represent the different location quotient values, making it easy to distinguish between higher and lower concentration compared with national average. It all clear, distinct representation of all the 51 locations easily. Here, value one is the National Average while values above or below are higher or lower respectively. 

### What we learned  

It shows that Rhode Island has a high concentration of Computer and Information Research Scientists with location quotient of 4.03 showing that the this particular occupation is 4.03 times as concetrated as the national average. It is followed by Maryland with 3.72  and Washington with 3.63. You can also see that most of the states have a location quotient below 1 indicating that they have lower occupational concentration for the job than National Average.  

The District of Columbia has the highest conectration Data Scientists with 3.48. Almost all of the states have a location quotient of just above one or below one indicating that the majority states have nealy an average or below average concentration of Data Scientists.  

When it comes ro Software Developers Washington, D.C leads with 2.34 followed by Virginia with 2.00. In this as well almost all states have an average or below average concentration than the National Average.

If we look at states, California has above-average concentartion across all states. Washington has the highest concentration for all occupation across all states. It is followed by Rhode Island, Utah and District Of Columbia.  

Overall, occupational concentration varies between geographical areas and occupations. Certain areas have strong concentration for computing occupations. This shows USA's differences in geographical distribution of computing workforce.

## **4. Challenge**  

### Compare the employment data across the years

The visualisations earlier was focused on the wahe and employment of USA in May 2023. The analysis is being expanded as employment data from 2022 and 2021 will be used to compare with 2023 to see how employment changed over time.  

The dataset from May 2022 and 2021 contains the same employment data as May 2023 with the same columns. The comparison will be made for the three selected technologies, Software Developers, Data Scientists and Computer and Information Research Scientists.  

The visualisation comparison will provide an insight into how the employment trends changed over the years form 2021 to 2023. It will display the trends of these selected occupations. 

### How employment in AI, Data Science and Software Development changed from 2021 to 2023 on a national level

### 4.1 Prepare the National Level data for this line chart

```python
# Select the occupations
selected_codes = [
    "15-1221", # Computer and Information Research Scientists
    "15-2051", # Data Scientists
    "15-1252" # Software Developers
]
```

```python
# Extract national cross-industry records for the selected occupations.
def extract_national_data(df, year):

    national_data = df[
        (df["AREA_TYPE"] == 1) &
        (df["I_GROUP"] == "cross-industry") &
        (df["OWN_CODE"] == 1235) &
        (df["O_GROUP"] == "detailed") &
        (df["OCC_CODE"].isin(selected_codes))
    ][[
        "OCC_CODE",
        "OCC_TITLE",
        "TOT_EMP",
        "A_MEDIAN"
    ]].copy()

    national_data["YEAR"] = year

    return national_data
```

```python
# Read 2021 data
df_2021 = pd.read_excel("all_data_M_2021.xlsx")

national_2021 = extract_national_data(
    df_2021,
    2021
)
```

```python
# Read 2022 data
df_2022 = pd.read_excel("all_data_M_2022.xlsx")

national_2022 = extract_national_data(
    df_2022,
    2022
)
```

```python
# Read data 2023
national_2023 = extract_national_data(
    df,
    2023
)
```

```python
# Checking if done right
national_2023
```

```python
# Combine the years into a single dataset
national_trend = pd.concat(
    [
        national_2021,
        national_2022,
        national_2023
    ],
    ignore_index = True
)
```

```python
#Convert wages and employment to numeric
national_trend["TOT_EMP"] =pd.to_numeric(
     national_trend["TOT_EMP"],
     errors = "coerce"
)

national_trend["A_MEDIAN"] = pd.to_numeric(
    national_trend["A_MEDIAN"],
    errors = "coerce"
)
```

```python
# Check for missing values
print(
    national_trend[
        ["TOT_EMP", "A_MEDIAN"]
        ].isna().sum()
)
```

```python
# Create a clean copy by removing missing employment values
national_trend_clean = national_trend.dropna(
    subset = ["TOT_EMP"]
).copy() 
```

```python
# Verify the data
print(national_trend_clean)

print("\n\nShape:")
print(national_trend_clean.shape)

print("\n\nRecords per occupation:")
print(
    national_trend_clean["OCC_TITLE"]
    .value_counts()
)
```

```python
# Check the info
national_trend_clean.info()
```

### 4.2 Median Annual Wage Trends From 2021 to 2023

```python
# Create a line chart showing the median annual wage trends from 2021 to 2023
plt.figure(figsize = (10,6))

sns.lineplot(
    data = national_trend_clean,
    x = "YEAR",
    y = "A_MEDIAN",
    hue = "OCC_TITLE",
    marker = "o",
    linewidth = 2.5
)

plt.title(
    "Median Annual Wage Trends (2021 - 2023)",
    fontsize = 15,
    fontweight = "bold"
)

plt.xlabel("Year")
plt.ylabel("Median Annual Wage ($)")

# Grid Lines
plt.grid(alpha = 0.3)

plt.tight_layout()
plt.show()
```

### Objective  

The objective of this Line Chart is to analyse how the mdeian annual wages changed in the years 2021, 2022, 2023 for the three selected occupations. The chart helps identify wage growth trends and compare salary progression.  

### Why Choose Line Chart  

A line chart is designed to display the changes over time. It makes it an appropriate chart for this visualisation. It connects wage values of 2021, 2022 and 2023 making it easy to identify the trends, growth patterns and differences between occupations. It also makes possible the comparison of multiple occupations simultaneously without cluttering the visualisation.  

### What We Learned  

The chart shows us that Computer and Information Research Scientists consistently earns the highest wages and has the highest median annual wages.It increases from 131,490 in 2021 to 145,080 in 2023.  

Software Developers have a steady wage growth across the years. They go from 120,730 in 2021 to 132,270 in 2023. Data Scientists has the lowest median wages amongst the three. It still shows consistent growth from 100,910 in 2021 to 108,020 in 2023.  

This shows that there is a high demand for highly skilled technology professionals during these years. Even though Data Scientists earn less thatn other, the steady growth indicates that the field will continue to grow in value in the coming years like the other two.
