When visualisation requires the columns and rows with these symbols they should be coverted, if not they should be replaced, dropped or dealt with. Except that the dataset will remain unchanged.

## **2.National Ranking Chart**  

### 2.1 Prepare data for National Level

```python
# Check data types to filter the National Values
print(df[["AREA_TYPE", "I_GROUP", "OWN_CODE", "O_GROUP"]].dtypes)  

print("Example category values:")
display(df[["AREA_TYPE", "I_GROUP", "OWN_CODE", "O_GROUP"]].head())
```

```python
# Create nat_detailed as a copy 
nat_detailed = df[
    (df["AREA_TYPE"] == 1) &
    (df["I_GROUP"] == "cross-industry") &
    (df["OWN_CODE"] == 1235) &
    (df["O_GROUP"] == "detailed") &
    (df["OCC_CODE"].str.startswith("15-", na=False))
    ].copy()
```

```python
# Select the required columns
nat_detailed = nat_detailed[
    ["OCC_CODE", "OCC_TITLE", "TOT_EMP"]
    ].copy()
```

```python
# Convert employment to numeric
nat_detailed["TOT_EMP"] = pd.to_numeric(
    nat_detailed["TOT_EMP"],
    errors="coerce"
)
```

```python
# Sort by employment
nat_detailed = nat_detailed.sort_values(
    "TOT_EMP",
    ascending=False
).reset_index(drop=True)
```

```python
# Verify the dataset
print("Dataset shape:", nat_detailed.shape)

print("Data Types:")
print(nat_detailed.dtypes)

print("Missing Values:")
print(nat_detailed.isna().sum())

print("Top 10 Occupations:")
print(nat_detailed.head(10))
```

### 2.2 National Employment Ranking Chart Across All Occupation Related To Computer And Maths

```python
# Highlighing the selected three occupation

selected_occupations = [
    "Computer and Information Research Scientists",
    "Data Scientists",
    "Software Developers"
]

bar_colors = [
    "#006684" if occupation in selected_occupations
        else "#B8C9CE"
            for occupation in nat_detailed["OCC_TITLE"]
]

# Create the chart
plt.figure(figsize=(13,10))

# Create the horizontal bars
bars = plt.barh(
    nat_detailed["OCC_TITLE"],
    nat_detailed["TOT_EMP"],
    color = bar_colors
)

# Display the highest employments at the top
plt.gca().invert_yaxis()

plt.title(
    "National Employment Rankings Of Computer And Mathematical Occupations",
    fontsize = 15,
    fontweight = "bold",
    pad= 18
)
plt.xlabel("Total Employment")
plt.ylabel("Occupation")

# Makes employment easy to compare
plt.grid(axis = "x", alpha = 0.25)

# Adding values to the bars
plt.bar_label(
    bars,
    labels=[
        f"{value:,.0f}"
        for value in nat_detailed["TOT_EMP"]
    ],
    padding = 4,
    fontsize = 8
)

# Prevents any labels from being cut off
plt.tight_layout()

plt.show()
```

### Objective  
The objective of this horizontal bar chart is to compare the National Employment Levels of 21 detailed Computer and Maths Occupations in the US. It also specifically highlights the ranking of the three selected occupations such as Data Scientists, Software Developers and Computer and Research Scientists. They are highlighted as they will be used for further analysis and visualisations.

### Why choose a Horizontal Bar  
It was selected as it allows easy comparison of employment figures across multiple occupations. The horizontal layout makes oit easier to accomodate longer names and also understanding the long list of occupational information easier. It makes volume differences for such long list of occupations easier and also does a better job at displaying them in descending order.  

### What We Learned  
The visualised data shows us that Software Developers have the highest employment among the 21 occupations with 1,656,880 making them rank first. Data Scientist holds the rank 7 with 192,710 workers. Computer and Research Scientist ranks 17th with 35,210 workers making them the one with least jobs among the selected occupations.  

The results suggests that there is a substancial difference in the employment level in these three areas. Software Developers have nearly 8.6 times jobs than Data Scientists and around 47 times that of Computer and Research Scientists. The US Employment Data of May 2023 shows us that the employment scale of Software Developers is much more than that of the other two selected occupations.

## **3. State-Level Occupation Charts**

### 3.1 Prepare data for State-Level

```python
# Define the three selected occupation with detailed
selected_codes = [
    "15-1221",
    "15-1252",
    "15-2051"
]
```

```python
# Filter state level records
state_data = df[
    (df["AREA_TYPE"] == 2) &
    (df["I_GROUP"] == "cross-industry") &
    (df["OWN_CODE"] == 1235)&
    (df["OCC_CODE"].isin(selected_codes))
    ].copy()
```

```python
# Selecting the required columns
state_data = state_data[
    [
        "AREA_TITLE",
        "PRIM_STATE",
        "OCC_CODE",
        "OCC_TITLE",
        "TOT_EMP",
        "JOBS_1000",
        "LOC_QUOTIENT",
        "A_MEDIAN"
    ]
    ].copy()
```

```python
# Convert the numerical columns to numeric to make sure everything is in numeric and to avoid any future issues
numeric_columns = [
    "TOT_EMP",
    "JOBS_1000",
    "LOC_QUOTIENT",
    "A_MEDIAN"
]

for col in numeric_columns:
    state_data[col] = pd.to_numeric(
        state_data[col],
        errors = "coerce"
    )
```

```python
# Verifying the filtered data
print("Dataset shape:")
print(state_data.shape)

print("\n\nData Types:")
print(state_data.dtypes)

print("\n\nMissing Values:")
print(state_data.isna().sum())

print("\n\nFirst 10 records:")
display(state_data.head(10))
```

```python
# Investigate the missing values
missing_records = state_data[
    state_data[
        ["TOT_EMP", "JOBS_1000", "LOC_QUOTIENT"]
        ].isna().any(axis=1)
    ]

display(
    missing_records[
        [
            "AREA_TITLE",
            "PRIM_STATE",
            "OCC_TITLE",
            "TOT_EMP",
            "JOBS_1000",
            "LOC_QUOTIENT"
        ]
        ]
)
```

#### State-Level Data Preparation
The above preparation of state-level data was filtered to include three of the selected occupations across all US States and the District of Columbia. Eight columns were retained for the charts.  

Missing Values were identified in the employment and concentration measures. All of these values were not replaced with zero because doing so could create issues with the visualisation accuracy. Instead, the missing records will be excluded when the relevant variable is used for visualisation.

### 3.2 Median Annual Wage Variation Across States

```python
# Prepare the chart data

# Remove records where the wage is missing
box_data = state_data.dropna(
    subset = ["A_MEDIAN"]
).copy()

print("Box plot dataset shape:", box_data.shape)

print("\n\nMissing median wages:")
print(box_data["A_MEDIAN"].isna().sum())
```

```python
# Create the chart
plt.figure(figsize = (11, 7))

sns.boxplot(
    data = box_data,
    x = "OCC_TITLE",
    y = "A_MEDIAN",
    hue = "OCC_TITLE",
    palette = {
        "Computer and Information Research Scientists": "blue",
        "Data Scientists": "green",
        "Software Developers": "orange"
    },
    legend = False
)

# Add the title
plt.title(
    "Variation in Median Annual Wages Across 50 States And Washington DC By Occupation",
    fontsize = 15,
    fontweight = "bold",
    pad = 15
)

# Label the axis
plt.xlabel("Occupation")
plt.ylabel("Median Annual Wage ($)")

# Rotate the labels to avoid overlap
plt.xticks(rotation = 15)
plt.grid(axis = "y", alpha = 0.25)

# Prevents any labels from being cut off
plt.tight_layout()

plt.show()
```

### Objective  
This visualisation examines how median annual wages varies across different states for Data Scientists, Software Developers and Computer and Information Research Scientists.  

### Why Use Box Plot  
A box plot was chosen it gives a clear picture of the wage distribution. It also allows the comparison of median, spread and outliers across the three occupations. It helps in understanding the differences in wage levels.  

### What We Learned  
It shows that the Computer and Information Research Scientists have the highest median annual wage, of approximately 121,000. It is folowed by Software Developers with 117,000. Data Scientists have tyhe lowest with 101,000.    

Computer and Information Research Scientists show a high-wage outlier at approximately 218,000. SDoftware Developers display a consistent wage distribution, while Data Scientists have a lower median wage.   

The visualisation shows the differences in both typical wages and geographical wage variation across the selected occupation

### 3.3 State Level Employment versus Median Annual Wages

```python
# Prepare the data for the chart

# Remove the missing median values
scatter_data = state_data.dropna(
    subset = ["TOT_EMP"]
).copy()

# Remove invalid values like zero or negative values
scatter_data = scatter_data[
    (scatter_data["TOT_EMP"] > 0) &
    (scatter_data["A_MEDIAN"] > 0)
    ].copy()

print("Scatter plot dataset shape:", scatter_data.shape)

print("\n\nMissing Values:")
print(scatter_data[["TOT_EMP", "A_MEDIAN"]].isna().sum())
```

```python
# Create the chart
plt.figure(figsize =(11, 7))

sns.scatterplot(
    data = scatter_data,
    x = "TOT_EMP",
    y = "A_MEDIAN",
    hue = "OCC_TITLE",
    style = "OCC_TITLE",
    s = 100,
    palette = {
        "Computer and Information Research Scientists": "blue",
        "Data Scientists": "green",
        "Software Developers": "red"
    }
)

# Stopping small values getting squeezed to the left
plt.xscale("log")

# Chart title
plt.title(
    "State Employment vs Median Annual Wages",
    fontsize = 15,
    fontweight = "bold",
    pad = 15
)

# Labelling the axises
plt.xlabel("Total Employment (Logarithmic Scale)")
plt.ylabel("Median Annual Wage ($)")

# Grid lines
plt.grid(alpha = 0.25)

# Shows which colour and marker corresponds to each occupation
plt.legend(title = "Occupation")

plt.tight_layout()
plt.show()
```

### Objective  
The objective of this visualisation ia to explore the relationship between the employment levels and median annual wages for the selected three occupations across the US geographical areas.  

### Why Use Scatter Plot  
A Scatter Plot was used because it allows two distinct numerical variables to be compared at the same time. Each point will represent a geographical area, while different colours and markers represent the three selected occupations. A logarithmic scale was applied to the employment axis of this scatter plot because employment scales varies a lot among the occupations.  

### What We Learned  
The chart shows us that Software Developers generally have higher employment levels compared to the other two occupations. Median Annual Wages ranges between 100,000 and 140,000.  

Computer and Information Research Scientists generally have lower employment among the three but has higher median wages. They vary greatly including one observation going upto 218,000.  

Data Scientists falls between the above two on the people employed. They show variation in median wages across multiple areas, while most of it fall between 80,000 and 130,000.  

Overall, it shows that employment levels and median annual wages vary across occupations and geographical areas. However, it does not show a consistent relationship more employment and higher median wages.

### 3.4 Software Developers' Median Wage Distribution

```python
# Select only software Developers and remove missing values
software_data = state_data[
    state_data["OCC_CODE"] == "15-1252"
    ].dropna(
        subset = ["A_MEDIAN"]
    ).copy()

print("Histogram dataset shape:", software_data.shape)

print("\n\nMinimum median wage:")
print(software_data["A_MEDIAN"].min())

print("\n\nMaximum median wage:")
print(software_data["A_MEDIAN"].max())
```

```python
# Create the chart
plt.figure(figsize = (11, 7))

sns.histplot(
    data = software_data,
    x = "A_MEDIAN",
# Group wages to interval bins
    bins = 12,
# Smooth density curve over the bars
    kde = True,
    color = "#006684"
)

plt.title(
    "Distribution of State Median Annual Wages for Software Developers",
    fontsize = 15,
    fontweight = "bold",
    pad = 15
)

plt.xlabel("Median Annual Wage ($)")
plt.ylabel("Number of Geographical Areas")

plt.grid(axis = "y", alpha = 0.25)

plt.tight_layout()
plt.show()
```

### Objective  

The objective is to examine the median annaul wages for Software Deevlopers across the 50 US States and Washington, D.C.  

### Why Choose A Histogram  

A histogram was selected because it allows grouping the numerical values into intervals. It makes understanding the frequency and distribution of wage estimates easier. It helps in identofying the common wage range, concentration of wages and overall spread of mdeian wages across the geographical areas. A KDE curve was also included to provide a smooth representation.  

### What We Learned  

It shows that most of the areas have a median wages between 100,000 and 140,000. Two concentrations to notice are around 105,000 - 115,000 and 125,000 - 135,000. The latter one has the highest frequency.  

Only a small number of areas have median wages above 150,000. The overall range spans from 80,000 to just over 160,000 for Software Developers. It shows the considerable variation in median wages across different geographical areas.  

As a whole, even though most areas fall within a particular concentration of median wage range, some of them have high median annual wage.
