## 1. Introduction

This project examines USA's employment statistics from May 2023. This dataset was taken from US Bureau of Labour Statistics, from its Occupational Employment and Wage Statistics (OEWS) Tables. This dataset contains information about multiple jobs across USA. It goes in detail about the Job Types, Mean Annual Wages, Median Annual Wages, Area, etc. It has 413,328 rows and 32 columns. In this CA Project, the main focus is three main areas:  

**- Artificial Intelligence**  
**- Data Science**  
**- Software Development**
